Dictionary

```python
student = {
    "name": "Rahul",
    "roll": 101,
    "marks": 87.5
}

print("Name:", student["name"])
print("Roll:", student["roll"])
print("Marks:", student["marks"])
```

In C,adding a new field to a struct requires changing the struct definition.What happens in Python when you add a new key to a dictionary while the program is running-and why is that both useful and risky?

Why is this useful?
It gives Python flexibility. You don't need to decide all the fields beforehand. If later you realize that you need to store age, email, or phone, you can simply add them.
Why is it risky?
The same flexibility can cause inconsistency and errors. Different dictionaries can have completely different keys.
Python gives a KeyError because that key doesn't exist.

The C program declares that marks is a float.The python dictionary does not declare anything.Where does the type come from?

In Python, the type comes from the value you assign, not from the dictionary itself.Python is dynamically typed: the type belongs to the object/value, and Python determines it at runtime.
